```python
#imports

import numpy as np
```

```python
#the given matrices

x_t = np.array( [[0.7,0.3],[0.3,0.7]])

e_t = np.array( [0.9, 0.2])
```

Using the Equation 14.5 $\newline$
 $P(X_{t+1} \mid e_{1:t+1}) = \alpha P(e_{t+1}\mid X_{t+1}) \sum_{X_t} P(X_{t+1}\mid x_t. e_{1:t})P(x_t \mid e_{1:t}) = \alpha P(e_{t+1}\mid X_{t+1}) \sum_{X_t} P(X_{t+1} \mid x_t) P (x_t \mid e_{1:t}) $

 Verify your implementation by calculating $P(X_2\mid e_{1:2})$, where $e_{1:2}$ is the evidence that the
umbrella was used both on day 1 and day 2. The desired result is that the probability of
rain at day 2 (after the observations) is 0.883.

```python
#function for forward step
def forward_step(belief, x_t, e_t, evidence):

    if evidence:           # umbrella = True
        obs = e_t
    else:                  # umbrella = False
        obs = (1 - e_t)

    belief = x_t.T @ belief      # prediction
    belief = obs * belief        # update
    belief /= np.sum(belief)     # normalize
    return belief
```

```python
belief = np.array([0.5, 0.5]) #day one

belief = forward_step(belief, x_t, e_t, True)  # day 1
belief = forward_step(belief, x_t, e_t, True)  # day 2

print(f'The probability of rain at day 2 given the observations is {belief[0]:.3f}')
```

Use your program to calculate the probability of rain at day 5 given the following sequence
of observations:

$e_{1:5} = {Umbrella_1 = true,\space \space  Umbrella_2 = true,
\space \space  Umbrella_3 = false, \space \space 
Umbrella_4 = true, \space \space  Umbrella_5 = true}.$

Document your answer by showing all $\textit{normalized}$ forward messages (in the book, the
un-normalized forward messages are denoted $\mathbf{f}_{1:k}$ for k = 1, 2, . . . , 5) in the PDF report.

```python
#defining a lst of the given observations
observations = [True, True, False, True, True]

#an empty list with the forward messages
forward_messages = []

#important to resset the belief
belief = np.array([0.5, 0.5]) #day one

 #looping through the observations and calculating the forward messages
for obs in observations:
    belief = forward_step(belief, x_t, e_t, obs)
    forward_messages.append(belief.copy())  

#printing the forward messages
for i in range(len(forward_messages)):
    msg = forward_messages[i]
    print(f"f_1:{i+1} = <{msg[0]:.3f}, {msg[1]:.3f}>")


print(f'The probability of rain at day 5 given the observations is {belief[0]:.3f}')

```
