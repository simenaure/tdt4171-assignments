```python
import random
```

```python
def random_day():
    return random.randint(1, 365)
```

```python
N = 23
count = 0
trials = 20000




def birthday(N):
    birthdays = []
    for _ in range(N):
        birthdays.append(random_day())
        if len(set(birthdays)) < len(birthdays):
            return True
    return False
        


for _ in range(trials):
    if birthday(N):
        count += 1

p_hat = count / trials
print("P(shared birthday) =", p_hat)
```

```python
N_values = list(range(10, 51))


for N in N_values:
    count = 0
    for _ in range(trials):
        if birthday(N):
            count += 1
    p_hat = count / trials
    print(N, p_hat)
```

```python
#Find number of people needed to cover every birthday
def every_birthday():
    birthdays = []
    while len(set(birthdays)) < 365:
            birthdays.append(random_day())
    
    return len(birthdays)


print("Number of people needed for every birthday:",every_birthday())
            
```
