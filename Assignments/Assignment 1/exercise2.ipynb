```python
import numpy as np
import random 
```

```python
symbols = ["CHERRY", "LEMON", "BELL", "BAR"]

#Function to spinn the slot machine
def spin_reels():
    return [random.choice(symbols) for _ in range(3)]
```

```python
#Function to calculate payback 
def payout(reels):
    if reels[0] == "BAR" and reels[1] == "BAR" and reels[2] == "BAR":
        return 20
    elif reels[0] == "BELL" and reels[1] == "BELL" and reels[2] == "BELL":
        return 15
    elif reels[0] == "LEMON" and reels[1] == "LEMON" and reels[2] == "LEMON":
        return 5
    elif reels[0] == "CHERRY" and reels[1] == "CHERRY" and reels[2] == "CHERRY":
        return 3
    elif reels[0] == "CHERRY" and reels[1] == "CHERRY":
        return 2
    elif reels[0] == "CHERRY":
        return 1
    else:
        return 0
```

```python
def play(start_cash):
    cash = start_cash
    rounds = 0
    
    while cash>0:
        reels = spin_reels()
        win = payout(reels)
        cash +=  - 1 + win
        rounds += 1
        
    return rounds
    

#Number of simulations
N = 10000

all_rounds = []

for _ in range(N):
    all_rounds.append(play(10))

mean_rounds = np.mean(all_rounds)
median_rounds = np.median(all_rounds)

print("Mean number of rounds", mean_rounds)
print("Median number of rounds", median_rounds)
```
